```python
df = patients.merge(appointments, on="patient_id", how="left")
```

```python
df = df.merge(treatments, on="appointment_id", how="left")
```

```python
df = df.merge(billing, on="patient_id", how="left")
```

```python
df.head()
df.isna().sum()
```

```python
df = df.dropna(subset=['amount'])
```

```python
df['treatment_type'] = df['treatment_type'].fillna("No Treatment")
df['status'] = df['status'].fillna("Unknown")
```

```python
df['cost'] = df['cost'].fillna(0)
df['amount'] = df['amount'].fillna(0)
```
